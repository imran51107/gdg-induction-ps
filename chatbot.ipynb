```python
dicticker={"apple":"AAPL","microsoft":"MSFT","tesla":"TSLA","google":"GOOG"}
```

```python
def extract_ticker(usinput):               
    usinput.lower()
    for name,ticker in dicticker.items():
        if name in usinput:
            return ticker
            
```

```python
def intent_detect(usinput):
    usinput=usinput.lower()
    if "why" in usinput and ("drop" in usinput or "fail" in usinput or "fall" in usinput or "falling" in usinput or "droppinig" in usinput or "failing" in usinput or "decrease" in usinput or "decreasing" in usinput) :
        intent="explain_drop"
    elif "when" in usinput and ("up" in usinput or "rise" in usinput or "increase" in usinput or "rising" in usinput or "increasing" in usinput):
        intent="find_uptrend"
    elif ("forecast" in usinput  or "predict" in usinput):
        intent="forecast"
    elif "trend" in usinput:
        intent="trend"
    else:
        intent="unknown"
    return intent
```

```python
usinput=input("ask:")
ticker=extract_ticker(usinput)
print("detected ticker:",ticker)
intent=intent_detect(usinput)
print("detected intent:",intent)
```

```python
import pandas as pd
```

```python
def getstockdata(ticker):
    data=yf.download(ticker,period="1y")
    return data
```

```python
data=getstockdata("AAPL")
data.head()
```

```python
def find_uptrend(ticker):
    data=getstockdata(ticker)
    data["daily_change"]=data["Close"].diff()
    max_increase=data["daily_change"].max()
    date=data["daily_change"].idxmax()
    return f"The biggest rise inn {ticker} was on {date.date()} with an increase of {max_increase:.2f}."
```

```python
print(find_uptrend("AAPL"))
```

```python
print(data.columns)
```

```python
def explain_drop(ticker):
    data = getstockdata(ticker)
    data["pct_change"] = data["Close"].pct_change() * 100
    worst_drop = data["pct_change"].min()
    date = data["pct_change"].idxmin()
    return (
        f"{ticker} experienced its biggest drop on {date.date()} "
        f"with a decline of {abs(worst_drop):.2f}% in a single day."
    )
```

```python
print(explain_drop("AAPL"))
```

```python
while True:
    usinput = input("Ask about a stock (type 'exit' to quit): ")
    if usinput.lower() == "exit":
        print("Goodbye!")
        break
    ticker = extract_ticker(usinput)
    intent = intent_detect(usinput)
    if ticker is None:
        print("please try again with different one")
        continue
    if intent == "find_uptrend":
        print(find_uptrend(ticker))
    elif intent == "explain_drop":
        print(explain_drop(ticker))
    else:
        print("ask me about stocks ")
```
